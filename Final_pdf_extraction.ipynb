```python
# Install the necessary libraries
!pip install pdfplumber pandas

import pdfplumber
import pandas as pd

# Define the input PDF file path
pdf_path = "/content/reed switch data sheet.pdf"

# Open the PDF with pdfplumber
with pdfplumber.open(pdf_path) as pdf:
    # Initialize an empty list to store data from all tables
    combined_data = []

    # Loop through all pages of the PDF
    for page_num, page in enumerate(pdf.pages, start=1):
        # Extract tables from the page
        tables = page.extract_tables()

        # If tables exist, process each table
        for table in tables:
            # Convert the table to a DataFrame
            df = pd.DataFrame(table)
            # Add a column to track page and table info (optional)
            df['Page'] = page_num
            combined_data.append(df)

# Combine all tables into a single DataFrame
combined_df = pd.concat(combined_data, ignore_index=True)

# Save the combined data to a CSV file
csv_path = "/content/extracted_tables.csv"
combined_df.to_csv(csv_path, index=False)

print(f"All tables have been extracted and saved to: {csv_path}")
from google.colab import files
files.download('/content/extracted_tables.csv')
```
